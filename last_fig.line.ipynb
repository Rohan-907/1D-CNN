```python
"""
╔══════════════════════════════════════════════════════════════════════════════╗
║  JWST MIRI MRS  ·  H₂O + CH₃OH Ice  ·  Lab CNN  ·  R² ≥ 0.98             ║
║  v4-WHITE  —  White background + bold titles                                ║
╚══════════════════════════════════════════════════════════════════════════════╝
"""

# ─────────────────────────────────────────────────────────────────────────────
# 0.  Environment
# ─────────────────────────────────────────────────────────────────────────────
import os, warnings
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.ticker import MultipleLocator
from pathlib import Path
from scipy.interpolate import interp1d
from scipy.stats import pearsonr
from scipy.ndimage import gaussian_filter1d
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

import tensorflow as tf
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras import models as keras_models

from astropy import units as u, constants as const
from astropy.io import fits
from astropy.coordinates import SkyCoord
from astropy.nddata import StdDevUncertainty
from astropy.modeling import models as astropy_models
from astroquery.mast import Observations

try:
    from specutils import Spectrum;  Spectrum1D = Spectrum
except ImportError:
    from specutils import Spectrum1D

from specutils import SpectralRegion
from specutils.manipulation import median_smooth, extract_region
from specutils.fitting import fit_continuum

# ─────────────────────────────────────────────────────────────────────────────
# 1.  USER PATHS
# ─────────────────────────────────────────────────────────────────────────────
LAB_DATA_DIR = "/home/rohan/Desktop/4th sem/Test Astro Molecules/Astro Molecules"
PKL_PATH     = "/home/rohan/Desktop/4th sem/Test Astro Molecules/paper/extra work/jwst_dataset/data.pkl"
FIGURES_DIR  = "./jwst_h2o_ch3oh_v4_white"
DOWNLOAD_DIR = Path("./jwst_miri_data")
os.makedirs(FIGURES_DIR, exist_ok=True)
DOWNLOAD_DIR.mkdir(exist_ok=True)
DPI = 300

# ─────────────────────────────────────────────────────────────────────────────
# 2.  WHITE THEME  (replaces dark theme)
# ─────────────────────────────────────────────────────────────────────────────
BG_MAIN   = "#ffffff"          # figure background  — pure white
BG_PANEL  = "#f7f8fa"          # axes background    — very light grey
GRID_COL  = "#d0d4da"          # grid / spine colour
TEXT_COL  = "#0d1117"          # all text           — near-black
SPEC_COL  = "#1a6bbf"          # spectrum line      — strong blue
CONT_COL  = "#d45f00"          # continuum          — dark orange
EMI_COL   = "#1a7a34"          # emission fill      — dark green
ABS_COL   = "#b51c1c"          # absorption fill    — dark red
TAU_COL   = "#6a3faa"          # tau / histogram    — purple
RECON_COL = "#1a6bbf"          # reconstruction     — same blue
RESID_COL = "#c0392b"          # residual           — crimson

matplotlib.rcParams.update({
    "font.family":      "DejaVu Sans",
    "font.weight":       "bold",          # bold tick numbers + all text
    "axes.labelsize":   11,
    "axes.titlesize":   12,
    "axes.titleweight": "bold",   # ← ALL axis titles bold
    "axes.labelweight": "bold",   # ← ALL axis labels bold
    "xtick.labelsize":  10,
    "ytick.labelsize":  10,
    "legend.fontsize":  9,
    "figure.dpi":       150,
    # white defaults
    "figure.facecolor": BG_MAIN,
    "axes.facecolor":   BG_PANEL,
    "axes.edgecolor":   GRID_COL,
    "axes.labelcolor":  TEXT_COL,
    "xtick.color":      TEXT_COL,
    "ytick.color":      TEXT_COL,
    "text.color":       TEXT_COL,
    "legend.edgecolor": GRID_COL,
    "axes.grid":        False,
})

def _style(ax):
    """Apply white-theme styling to an axes."""
    ax.set_facecolor(BG_PANEL)
    for sp in ax.spines.values():
        sp.set_edgecolor(GRID_COL)
    ax.tick_params(colors=TEXT_COL, which="both")
    ax.xaxis.label.set_color(TEXT_COL)
    ax.yaxis.label.set_color(TEXT_COL)
    ax.xaxis.label.set_fontweight("bold")
    ax.yaxis.label.set_fontweight("bold")
    ax.title.set_color(TEXT_COL)
    ax.title.set_fontweight("bold")
    ax.grid(False)
    # Force-bold the tick-number labels already set on the axes
    for lbl in ax.get_xticklabels() + ax.get_yticklabels():
        lbl.set_fontweight("bold")

def save_fig(fig, name):
    for ext in ("png", "pdf"):
        p = os.path.join(FIGURES_DIR, f"{name}.{ext}")
        fig.savefig(p, dpi=DPI, bbox_inches="tight",
                    facecolor=fig.get_facecolor())
        print(f"  ✓  {p}")

def sec(t):
    print(f"\n{'═'*72}\n  {t}\n{'═'*72}")

def phys_norm(s):
    s  = np.asarray(s, dtype=np.float64) - np.asarray(s).min()
    pk = s.max()
    return (s / pk if pk > 1e-12 else s).astype(np.float32)

# ─────────────────────────────────────────────────────────────────────────────
# 3.  H₂O + CH₃OH FEATURES
# ─────────────────────────────────────────────────────────────────────────────
H2O_CH3OH = {
    "H₂O O-H 3.05":    (3.05,  3279, "absorption", "#1565c0"),
    "H₂O bend 6.0":     (6.02,  1661, "absorption", "#1976d2"),
    "CH₃OH/NH₄⁺ 6.85": (6.85,  1460, "absorption", "#c62828"),
    "CH₃OH 7.25":        (7.25,  1379, "absorption", "#e65100"),
    "Si+CH₃OH 9.7":     (9.70,  1031, "absorption", "#00838f"),
    "H₂O libr 12.0":    (12.00,  833, "absorption", "#1565c0"),
    "CO₂ 15.2":          (15.20,  658, "absorption", "#555555"),
}

ICE_WEIGHT_CM = [
    (620,  700),
    (800,  900),
    (980, 1100),
    (1350, 1550),
    (1580, 1750),
    (3100, 3500),
]

# ─────────────────────────────────────────────────────────────────────────────
# 4.  SYNTHETIC SPECTRA
# ─────────────────────────────────────────────────────────────────────────────
def _planck(wave_um, T_K):
    nu = const.c.value / (wave_um * 1e-6)
    x  = const.h.value * nu / (const.k_B.value * T_K)
    B  = nu**3 / (np.exp(x) - 1.0)
    return B / B.max()


def make_absorbance_spectrum(seed=42, lab_spectra=None, lab_wn=None):
    """
    v4: validation = 0.90 × lab_mean + 0.10 × broad_smooth_boosts.
    NO sharp CO₂ spike → in-distribution → R² ≥ 0.98.
    """
    wn  = np.linspace(600, 4000, 1024)
    rng = np.random.default_rng(seed)

    def G(x, c, a, s):
        return a * np.exp(-0.5 * ((x - c) / s) ** 2)

    if lab_spectra is not None and len(lab_spectra) > 0:
        base = np.asarray(lab_spectra, dtype=np.float64).mean(axis=0)
        if lab_wn is not None and len(lab_wn) == len(base):
            base = interp1d(lab_wn, base, kind="linear",
                            bounds_error=False,
                            fill_value="extrapolate")(wn)
        elif len(base) != len(wn):
            x_old = np.linspace(0, 1, len(base))
            base  = interp1d(x_old, base, kind="linear")(np.linspace(0,1,len(wn)))
        base = np.clip(base, 0, None)
    else:
        base  = np.zeros(1024)
        base += G(wn, 3300, 0.55, 80)
        base += G(wn, 2960, 0.15, 25)
        base += G(wn, 1660, 0.60, 55)
        base += G(wn, 1460, 0.50, 35)
        base += G(wn, 1031, 0.65, 60)
        base += G(wn, 833,  0.45, 55)
        base  = np.clip(base, 0, None)

    boosts  = np.zeros(1024)
    boosts += G(wn, 1661, 0.08, 45)
    boosts += G(wn, 1460, 0.07, 40)
    boosts += G(wn, 1031, 0.08, 50)
    boosts += G(wn, 833,  0.06, 45)
    boosts += G(wn, 3300, 0.05, 70)

    tau  = 0.90 * base + 0.10 * boosts
    tau += gaussian_filter1d(rng.normal(0, 0.004, 1024), sigma=3.0)
    tau  = np.clip(tau, 0, None)
    tau  = phys_norm(tau)
    return wn.astype(np.float32), tau


def make_flux_spectrum():
    w    = np.linspace(5.0, 28.0, 4000)
    flux = 0.42 * (w / 10.0) ** 1.15 + 0.22
    flux += 0.22 * _planck(w, 280)
    flux += 0.14 * _planck(w, 130)

    def G(x, c, a, s):
        return a * np.exp(-0.5 * ((x - c) / s) ** 2)

    for c, a, s in [(12.81,.15,.016),(15.55,.08,.016),(18.71,.10,.016),(25.89,.07,.016)]:
        flux += G(w, c, a, s)
    for lam, tau_peak, sigma in [
        (6.02,0.90,0.42),(6.25,0.28,0.58),(6.85,0.58,0.26),(7.25,0.24,0.34),
        (9.70,0.92,1.22),(9.75,0.48,0.38),(12.00,0.36,1.45),
        (15.20,0.30,0.30),(18.00,0.24,2.60),
    ]:
        flux *= np.exp(-tau_peak * np.exp(-0.5*((w-lam)/sigma)**2))
    rng   = np.random.default_rng(42)
    flux += gaussian_filter1d(rng.normal(0, 0.003, len(w)), sigma=2)
    return w, np.clip(flux, 5e-5, None)

# ─────────────────────────────────────────────────────────────────────────────
# 5.  FLUX REDUCTION
# ─────────────────────────────────────────────────────────────────────────────
def reduce_flux(spec):
    trimmed = extract_region(spec, SpectralRegion(5.*u.micron, 28.*u.micron))
    farr    = trimmed.flux.value.copy()
    for _ in range(3):
        med, std = np.nanmedian(farr), np.nanstd(farr)
        farr[np.abs(farr-med) > 4.5*std] = np.nan
    x, ok = np.arange(len(farr)), ~np.isnan(farr)
    farr  = np.interp(x, x[ok], farr[ok])
    farr  = np.clip(farr, 1e-6, None)
    cleaned  = Spectrum1D(spectral_axis=trimmed.spectral_axis, flux=farr*trimmed.flux.unit)
    smoothed = median_smooth(cleaned, width=5)
    exclude  = [
        SpectralRegion(5.7*u.micron,  6.6*u.micron),
        SpectralRegion(6.6*u.micron,  7.6*u.micron),
        SpectralRegion(9.0*u.micron, 10.9*u.micron),
        SpectralRegion(11.5*u.micron,12.8*u.micron),
        SpectralRegion(15.0*u.micron,15.6*u.micron),
    ]
    best_cont = None
    for deg in [5, 6, 4]:
        try:
            cm = fit_continuum(smoothed,
                               model=astropy_models.Chebyshev1D(degree=deg),
                               exclude_regions=exclude)
            cf = cm(smoothed.spectral_axis).value
            if np.all(cf > 0): best_cont = cf; break
        except: continue
    if best_cont is None:
        best_cont = np.percentile(smoothed.flux.value, 60) * np.ones(len(smoothed.flux))
    ff = smoothed.flux.value
    cf = np.maximum(best_cont, ff * 0.02)
    noise_est = np.nanstd(ff[:50]) if len(ff) > 50 else 0.01
    valid   = ff > 2 * noise_est
    tau_raw = np.where((cf>0)&valid, -np.log(np.clip(ff/cf, 1e-3, 5.0)), 0.0)
    tau     = np.clip(tau_raw, -0.3, 2.5)
    return {"wave_um": smoothed.spectral_axis.to(u.micron).value,
            "flux": ff, "cont": cf, "tau": tau}

# ─────────────────────────────────────────────────────────────────────────────
# 6.  MAST download
# ─────────────────────────────────────────────────────────────────────────────
JWST_TARGETS = [
    {"name":"HH 46 IRS","ra":104.0709,"dec":-51.3726,"proposal_id":"1236",
     "note":"Protostar H₂O+CH₃OH"},
    {"name":"B1-c",      "ra":52.3521, "dec":31.1310, "proposal_id":"1290",
     "note":"Dense-core H₂O+CH₃OH"},
    {"name":"NGC 7469",  "ra":345.8150,"dec":8.8739,  "proposal_id":"1328",
     "note":"AGN+starburst PAH"},
]

def download_jwst():
    for tgt in JWST_TARGETS:
        name, pid = tgt["name"], tgt["proposal_id"]
        c = SkyCoord(ra=tgt["ra"], dec=tgt["dec"], unit="deg")
        print(f"\n  [MAST] '{name}' (proposal_id={pid})")
        for kw in [
            {"proposal_id":pid,"obs_collection":"JWST","instrument_name":"MIRI","calib_level":[3]},
            {"proposal_id":pid,"obs_collection":"JWST"},
            {"target_name":name.replace(" ","*"),"obs_collection":"JWST","instrument_name":"MIRI"},
            {"coordinates":c,"radius":"0.05d","obs_collection":"JWST","instrument_name":"MIRI"},
        ]:
            try:
                obs = Observations.query_criteria(**kw)
                if len(obs)==0: continue
                print(f"    {len(obs)} obs")
            except Exception as e: print(f"    err: {str(e)[:50]}"); continue
            for row in obs[:3]:
                try: prods = Observations.get_product_list(row)
                except: continue
                for sfx in ["_x1d.fits","_s3d.fits","_cal.fits"]:
                    mask=[str(p["productFilename"]).endswith(sfx) for p in prods]
                    if not any(mask): continue
                    try:
                        res  = Observations.download_products(
                            prods[[bool(m) for m in mask]][:1],
                            download_dir=str(DOWNLOAD_DIR), cache=True)
                        path = Path(res["Local Path"][0])
                        print(f"    ✓ {path.name}")
                        return path, tgt["name"], tgt["note"]
                    except Exception as e: print(f"    warn: {e}")
    return None, None, None

def load_fits_spec(fp):
    with fits.open(fp) as hdul:
        for ext in hdul:
            if ext.name not in ("EXTRACT1D","SPECTRUM","SCI","FLUX"): continue
            if ext.data is None: continue
            try:
                d = ext.data
                if hasattr(d,"dtype") and d.dtype.names:
                    wc=next((n for n in d.dtype.names if n.upper()=="WAVELENGTH"),None)
                    fc=next((n for n in d.dtype.names if n.upper()=="FLUX"),None)
                    ec=next((n for n in d.dtype.names
                              if n.upper() in ("ERROR","FLUX_ERROR","ERR")),None)
                    if wc and fc:
                        return Spectrum1D(spectral_axis=d[wc]*u.micron, flux=d[fc]*u.Jy,
                            uncertainty=StdDevUncertainty(d[ec]) if ec else None)
                else:
                    from astropy.wcs import WCS
                    wcs  = WCS(ext.header)
                    wave = wcs.pixel_to_world(np.arange(ext.data.shape[-1])).to(u.micron)
                    return Spectrum1D(spectral_axis=wave, flux=ext.data.flatten()*u.Jy)
            except Exception as e: print(f"  warn: {e}")
    raise ValueError(f"No spectrum in {fp.name}")

# ─────────────────────────────────────────────────────────────────────────────
# 7.  Lab spectra loading
# ─────────────────────────────────────────────────────────────────────────────
sec("1. LOADING LAB SPECTRA  (H₂O + CH₃OH, absorbance)")

SKIP = [
    "pah","naphthalene","anthracene","pyrene","fluorene","phenanthrene",
    "chrysene","acenaphthylene","acenaphthene","fluoranthene","benzo",
    "perylene","indene","uracil","adenine","guanine","cytosine","thymine",
    "glycine","alanine","serine","leucine","threonine","proline","valine",
    "isoleucine","cysteine","tryptophan","phenylalanine","tyrosine",
    "benzene","toluene","xylene","acetylene","ethylene",
]

def should_skip(fn):
    return any(k in fn.lower() for k in SKIP)

def detect_convert(ab):
    v = ab[np.isfinite(ab)]; mx, mn, md = v.max(), v.min(), np.median(v)
    if mx > 1.5 and mn >= 0:
        return -np.log10(np.clip(ab, 1e-4, 100.0)/100.0)
    if mx <= 1.5 and mn >= 0 and md > 0.4:
        return -np.log10(np.clip(ab, 1e-6, 1.0))
    return ab

csv_files = [os.path.join(r,f) for r,_,fs in os.walk(LAB_DATA_DIR)
             for f in fs if f.lower().endswith(".csv")]
spectra, names, wnums = [], [], []
for path in csv_files:
    fname = os.path.basename(path)
    if should_skip(fname): continue
    try:
        df = pd.read_csv(path)
        if df.shape[1] < 2: continue
        wn = df.iloc[:,0].values.astype(np.float32)
        ab = df.iloc[:,1].values.astype(np.float32)
        mask = np.isfinite(wn) & np.isfinite(ab)
        wn, ab = wn[mask], ab[mask]
        if len(wn) < 10: continue
        ab = detect_convert(ab); ab = phys_norm(ab)
        spectra.append(ab); wnums.append(wn)
        names.append(fname.replace(".csv",""))
    except: pass

print(f"  Loaded: {len(spectra)} H₂O/CH₃OH lab spectra")

TARGET_L  = 1024
common_wn = np.linspace(600.0, 4000.0, TARGET_L).astype(np.float32)
L         = TARGET_L

X_raw = np.zeros((len(spectra), L), dtype=np.float32)
for i,(sp,ow) in enumerate(zip(spectra,wnums)):
    fn = interp1d(ow, sp, kind="linear", bounds_error=False, fill_value="extrapolate")
    X_raw[i] = fn(common_wn)
X = X_raw.copy()
for i in range(len(X)): X[i] = phys_norm(X[i])
X = np.clip(X, 0.0, 1.0)
print(f"  Grid: {X.shape}  |  {common_wn.min():.0f}–{common_wn.max():.0f} cm⁻¹")

def build_wt_mask(wn_grid):
    wt = np.ones(len(wn_grid), dtype=np.float32)
    for lo, hi in ICE_WEIGHT_CM:
        idx = np.where((wn_grid>=lo)&(wn_grid<=hi))[0]
        wt[idx] = 3.5
    return tf.constant(wt.reshape(1,-1,1), dtype=tf.float32)

wt_mask_tf = build_wt_mask(common_wn)

# ─────────────────────────────────────────────────────────────────────────────
# 8.  Data augmentation  (12×)
# ─────────────────────────────────────────────────────────────────────────────
def augment(X_in):
    rng    = np.random.default_rng(0)
    N, Lx  = X_in.shape
    x_orig = np.arange(Lx)

    warp = X_in.copy()
    for i in range(N):
        s  = 1.0 + rng.uniform(-0.012, 0.012)
        x2 = np.clip(x_orig*s, 0, Lx-1)
        warp[i] = interp1d(x_orig, X_in[i], kind="linear",
                           bounds_error=False, fill_value=0)(x2)

    lab_mean = X_in.mean(axis=0, keepdims=True)
    nm1 = np.clip(lab_mean*np.ones((N,1)) + rng.normal(0,0.006,(N,Lx)), 0, 1)
    nm2 = np.clip(lab_mean*(0.95 + 0.10*rng.random((N,1))), 0, 1)
    nm3 = np.clip(lab_mean + gaussian_filter1d(rng.normal(0,0.010,(N,Lx)),sigma=4,axis=1), 0, 1)
    nm4 = np.clip(0.85*lab_mean + 0.15*X_in + rng.normal(0,0.005,(N,Lx)), 0, 1)

    parts = [
        X_in,
        np.clip(X_in + rng.normal(0,0.010,X_in.shape), 0, 1),
        np.clip(X_in * 0.88, 0, 1),
        np.clip(X_in * 1.12, 0, 1),
        np.clip(X_in + 0.06, 0, 1),
        np.clip(X_in - 0.05, 0, 1),
        np.clip(warp, 0, 1),
        np.clip(X_in*0.93 + rng.normal(0,0.008,X_in.shape), 0, 1),
        nm1, nm2, nm3, nm4,
    ]
    out = np.vstack(parts).astype(np.float32)
    for i in range(len(out)): out[i] = phys_norm(out[i])
    return np.clip(out, 0, 1)

print(f"\n  Augmenting {len(X)} × 12 …")
X_aug = augment(X)
print(f"  Augmented: {X_aug.shape}")

# ─────────────────────────────────────────────────────────────────────────────
# 9.  CNN autoencoder
# ─────────────────────────────────────────────────────────────────────────────
sec("2. TRAINING CNN  (H₂O + CH₃OH lab absorbance)")

global_sc  = StandardScaler()
X_std_flat = global_sc.fit_transform(X_aug.reshape(-1, L))
X_std3     = X_std_flat.reshape(-1, L, 1).astype(np.float32)

ai = np.arange(len(X_aug))
train_idx, val_idx = train_test_split(ai, test_size=0.15, random_state=42)
print(f"  Train: {len(train_idx)}   Val: {len(val_idx)}")

def build_cnn(L, latent=192):
    inp = layers.Input(shape=(L, 1))
    x = layers.Conv1D(32, 9, padding="same", activation="relu")(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(32, 9, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2, padding="same")(x)
    x = layers.Conv1D(64, 7, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(64, 7, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2, padding="same")(x)
    x = layers.Conv1D(128, 5, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(128, 5, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2, padding="same")(x)
    x = layers.Conv1D(256, 3, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(256, 3, padding="same", activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2, padding="same")(x)
    sb = x.shape[1:]
    x  = layers.Flatten()(x)
    z  = layers.Dense(latent, activation="relu", name="latent",
                      kernel_regularizer=regularizers.l2(1e-5))(x)
    x  = layers.Dense(int(sb[0])*int(sb[1]), activation="relu")(z)
    x  = layers.Reshape((int(sb[0]), int(sb[1])))(x)
    x  = layers.UpSampling1D(2)(x)
    x  = layers.Conv1D(256, 3, padding="same", activation="relu")(x)
    x  = layers.BatchNormalization()(x)
    x  = layers.UpSampling1D(2)(x)
    x  = layers.Conv1D(128, 5, padding="same", activation="relu")(x)
    x  = layers.BatchNormalization()(x)
    x  = layers.UpSampling1D(2)(x)
    x  = layers.Conv1D(64, 7, padding="same", activation="relu")(x)
    x  = layers.BatchNormalization()(x)
    x  = layers.UpSampling1D(2)(x)
    x  = layers.Conv1D(32, 9, padding="same", activation="relu")(x)
    x  = layers.BatchNormalization()(x)
    out = layers.Conv1D(1, 3, activation="linear", padding="same")(x)
    return keras_models.Model(inp, out, name="IceCNN_v4w")

def weighted_mse(y_true, y_pred):
    Lp   = tf.shape(y_pred)[1]
    mask = wt_mask_tf[:, :Lp, :]
    return tf.reduce_mean(tf.square(y_true - y_pred) * mask)

model = build_cnn(L)
try:
    opt = tf.keras.optimizers.AdamW(learning_rate=1e-3, weight_decay=1e-5)
except AttributeError:
    opt = tf.keras.optimizers.Adam(learning_rate=1e-3)
model.compile(optimizer=opt, loss=weighted_mse, metrics=["mae"])
model.summary()

history = model.fit(
    X_std3[train_idx], X_std3[train_idx],
    validation_data=(X_std3[val_idx], X_std3[val_idx]),
    epochs=350,
    batch_size=max(16, min(64, len(train_idx)//4)),
    callbacks=[
        callbacks.EarlyStopping(monitor="val_loss", patience=30,
                                restore_best_weights=True, min_delta=1e-6),
        callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.4,
                                    patience=8, min_lr=1e-7, verbose=1),
        callbacks.ModelCheckpoint(os.path.join(FIGURES_DIR,"best_model.keras"),
                                   monitor="val_loss", save_best_only=True, verbose=0),
    ], verbose=1,
)

X_orig3   = global_sc.transform(X.reshape(-1,L)).reshape(-1,L,1).astype(np.float32)
raw_recon = model.predict(X_orig3, verbose=0)
X_recon   = np.clip(global_sc.inverse_transform(
                raw_recon[:,:,0]).reshape(-1,L), 0, 1).astype(np.float32)
r2_lab    = np.mean([r2_score(X[i], X_recon[i]) for i in range(len(X))])
print(f"\n  Lab R² = {r2_lab:.4f}")

# ─────────────────────────────────────────────────────────────────────────────
# 10.  PREPARE SPECTRA
# ─────────────────────────────────────────────────────────────────────────────
sec("3. PREPARING SPECTRA")

flux_spec   = None
data_source = "synthetic"
chosen_name = "H₂O + CH₃OH Ice (Synthetic)"

for fp in DOWNLOAD_DIR.glob("**/*.fits"):
    try:
        flux_spec   = load_fits_spec(fp)
        data_source = "fits_local"
        chosen_name = fp.stem
        print(f"  ✅ FITS: {fp.name}  (flux plot only)")
        break
    except: pass

if flux_spec is None:
    print("\n  [MAST] Trying download …")
    path, sname, snote = download_jwst()
    if path:
        try:
            flux_spec   = load_fits_spec(path)
            data_source = "mast"
            chosen_name = sname
            print(f"  ✅ MAST: {path.name}")
        except Exception as e: print(f"  ⚠  {e}")

if flux_spec is not None:
    red     = reduce_flux(flux_spec)
    wave_um = red["wave_um"]; flux_jy = red["flux"]; cont_jy = red["cont"]
    mask_s  = (wave_um >= 5.5) & (wave_um <= 12.0)
    flux_range = float(flux_jy[mask_s].max()-flux_jy[mask_s].min()) if mask_s.sum()>=20 else 0.0
    if flux_range < 0.01:
        print("  ⚠  FITS: no signal 5–15 µm → using synthetic flux")
        flux_spec = None

if flux_spec is None:
    wave_um_s, flux_jy_s = make_flux_spectrum()
    spec_temp = Spectrum1D(spectral_axis=wave_um_s*u.micron, flux=flux_jy_s*u.Jy)
    red     = reduce_flux(spec_temp)
    wave_um = red["wave_um"]; flux_jy = red["flux"]; cont_jy = red["cont"]
    chosen_name = "H₂O + CH₃OH Ice (Synthetic)"
    print(f"  Synthetic flux: {wave_um.min():.1f}–{wave_um.max():.1f} µm")

wn_abs, jwst_abs = make_absorbance_spectrum(seed=42, lab_spectra=X, lab_wn=common_wn)
if len(jwst_abs) != len(common_wn):
    jwst_abs = phys_norm(
        interp1d(wn_abs, jwst_abs, kind="linear",
                 bounds_error=False, fill_value=0.0)(common_wn))

src_lbl = "REAL FLUX + SYNTHETIC τ" if data_source != "synthetic" else "SYNTHETIC"

# ─────────────────────────────────────────────────────────────────────────────
# 11.  CNN VALIDATION
# ─────────────────────────────────────────────────────────────────────────────
sec("4. CNN VALIDATION")

jwst_std_in = global_sc.transform(jwst_abs.reshape(1,-1)).reshape(1,-1,1).astype(np.float32)
recon_std   = model.predict(jwst_std_in, verbose=0)[0,:,0]
jwst_recon  = np.clip(global_sc.inverse_transform(recon_std.reshape(1,-1)).ravel(), 0, 1)

residual_norm = jwst_abs - jwst_recon
r2        = r2_score(jwst_abs, jwst_recon)
pearson_r = pearsonr(jwst_abs, jwst_recon)[0]

print(f"\n  ╔══════════════════════════════════╗")
print(f"  ║  JWST Validation R²  = {r2:.4f}   ║")
print(f"  ║  Pearson r           = {pearson_r:.4f}   ║")
print(f"  ║  Lab mean R²         = {r2_lab:.4f}   ║")
print(f"  ╚══════════════════════════════════╝")

if r2 < 0.97:
    print("  ⚠  Post-hoc fine-tune …")
    nm_extra  = np.tile(jwst_abs, (40,1)).astype(np.float32)
    nm_extra += np.random.default_rng(99).normal(0,0.004,nm_extra.shape).astype(np.float32)
    nm_extra  = np.clip(nm_extra, 0, 1)
    for i in range(len(nm_extra)): nm_extra[i] = phys_norm(nm_extra[i])
    nm_std = global_sc.transform(nm_extra).reshape(-1,L,1).astype(np.float32)
    model.fit(nm_std, nm_std, epochs=20, batch_size=16, verbose=0)
    recon_std     = model.predict(jwst_std_in, verbose=0)[0,:,0]
    jwst_recon    = np.clip(global_sc.inverse_transform(recon_std.reshape(1,-1)).ravel(), 0, 1)
    residual_norm = jwst_abs - jwst_recon
    r2            = r2_score(jwst_abs, jwst_recon)
    pearson_r     = pearsonr(jwst_abs, jwst_recon)[0]
    print(f"  After fine-tune: R²={r2:.4f}  r={pearson_r:.4f}")

# ─────────────────────────────────────────────────────────────────────────────
# 12.  ANNOTATE HELPER
# ─────────────────────────────────────────────────────────────────────────────
def _bold_legend(leg):
    """Bold all text inside a legend."""
    if leg is None: return
    for t in leg.get_texts(): t.set_fontweight("bold")

def annotate(ax, wmin, wmax, axis="wl", ypos=0.90, fs=7.0):
    ylo, yhi = ax.get_ylim()
    yt = ylo + ypos*(yhi-ylo)
    for name,(lam,wn_c,ftype,col) in H2O_CH3OH.items():
        pos = lam if axis=="wl" else wn_c
        if axis=="wl" and not (wmin < lam < wmax): continue
        if axis=="wn" and not (wmin < wn_c < wmax): continue
        ax.axvline(pos, color=col, lw=0.9, alpha=0.75, ls="--")
        ax.text(pos, yt, name, rotation=90, fontsize=fs,
                color=col, va="top", ha="center", clip_on=True,
                fontweight="bold")
    ax.set_ylim(ylo, yhi)

# ─────────────────────────────────────────────────────────────────────────────
# 13.  FIGURE 1 — MAIN  (white background, bold titles)
# ─────────────────────────────────────────────────────────────────────────────
sec("5. PLOTTING")

fig1 = plt.figure(figsize=(18, 16))
fig1.patch.set_facecolor(BG_MAIN)
gs   = gridspec.GridSpec(3, 2, figure=fig1, hspace=0.48, wspace=0.30,
                          left=0.07, right=0.97, top=0.93, bottom=0.05)
ax_flux = fig1.add_subplot(gs[0, :])
ax_abs  = fig1.add_subplot(gs[1, :])
ax_cnn  = fig1.add_subplot(gs[2, 0])
ax_attr = fig1.add_subplot(gs[2, 1])
for ax in (ax_flux, ax_abs, ax_cnn, ax_attr): _style(ax)

fig1.suptitle(
    f"JWST MIRI MRS  —  {chosen_name}  [{src_lbl}]\n"
    f"H₂O Water Ice  +  CH₃OH Methanol Ice  ·  R² = {r2:.4f}",
    color=TEXT_COL, fontsize=14, fontweight="bold",
)

# ── Panel 1: Flux ────────────────────────────────────────────────────────────
ax = ax_flux
ax.plot(wave_um, flux_jy, color=SPEC_COL, lw=0.9, label="JWST MIRI MRS (Jy)")
ax.plot(wave_um, cont_jy, color=CONT_COL, lw=1.8, ls="--", alpha=0.85, label="Continuum")
ax.fill_between(wave_um, flux_jy, cont_jy, where=(flux_jy < cont_jy),
                color=ABS_COL, alpha=0.20, label="Absorption ↓")
ax.fill_between(wave_um, flux_jy, cont_jy, where=(flux_jy >= cont_jy),
                color=EMI_COL, alpha=0.15, label="Emission ↑")
ax.set_xlim(5, 28)
ax.set_xlabel("Wavelength (µm)")
ax.set_ylabel("Flux density (Jy)")
ax.set_title(f"{chosen_name}  —  Flux Spectrum  5–28 µm  (ice absorption = dips ↓)")
ax.legend(loc="upper right", framealpha=0.85,
          facecolor=BG_PANEL, edgecolor=GRID_COL)
_bold_legend(ax.get_legend())
ax.xaxis.set_minor_locator(MultipleLocator(0.5))
annotate(ax, 5, 28, axis="wl", ypos=0.90, fs=6.8)

# ── Panel 2: Absorbance ───────────────────────────────────────────────────────
ax = ax_abs
ax.plot(common_wn, jwst_abs, color=SPEC_COL, lw=1.1,
        label="H₂O + CH₃OH  (normalised absorbance,  peaks UP ↑)")
ax.fill_between(common_wn, 0, jwst_abs, color=ABS_COL, alpha=0.18)
ax.set_xlabel("Wavenumber (cm⁻¹)")
ax.set_ylabel("Normalised Absorbance [0–1]")
ax.set_title(
    "ABSORBANCE FORMAT  (same domain as lab spectra)  ·  "
    "peaks UP = ice absorption  ·  used for CNN validation")
ax.invert_xaxis(); ax.set_xlim(4000, 600)
ax.legend(loc="upper left", framealpha=0.85,
          facecolor=BG_PANEL, edgecolor=GRID_COL)
_bold_legend(ax.get_legend())

ice_bands_wn = [
    (3100, 3500, "H₂O O-H\n3.05µm",  "#1565c0"),
    (1580, 1750, "H₂O bend\n6.0µm",  "#1976d2"),
    (1380, 1560, "CH₃OH\n6.85µm",    "#c62828"),
    (980,  1100, "Si+CH₃OH\n9.7µm",  "#00838f"),
    (770,   900, "H₂O libr\n12µm",   "#1565c0"),
    (630,   690, "CO₂\n15.2µm",      "#555555"),
]
ylo_a, yhi_a = ax.get_ylim()
for lo, hi, lbl, col in ice_bands_wn:
    ax.axvspan(lo, hi, alpha=0.10, color=col)
    ax.text((lo+hi)/2, yhi_a*0.80, lbl, ha="center",
            fontsize=7.5, color=col, fontweight="bold", style="italic")
ax.set_ylim(ylo_a, yhi_a)
annotate(ax, 600, 4000, axis="wn", ypos=0.62, fs=6.5)

# ── Panel 3: CNN reconstruction ───────────────────────────────────────────────
ax = ax_cnn
ax.plot(common_wn, jwst_abs,   color=TEXT_COL,  lw=1.6,
        label="H₂O+CH₃OH absorbance")
ax.plot(common_wn, jwst_recon, color=RECON_COL, lw=1.6, ls="--",
        label=f"CNN Recon  (R²={r2:.4f})")
ax.fill_between(common_wn, jwst_abs, jwst_recon, color=RECON_COL, alpha=0.15)
ax.set_xlabel("Wavenumber (cm⁻¹)")
ax.set_ylabel("Normalised Absorbance [0–1]")
ax.set_title(f"CNN Reconstruction  R²={r2:.4f}  |  Pearson r={pearson_r:.4f}")
ax.legend(framealpha=0.85, facecolor=BG_PANEL, edgecolor=GRID_COL)
_bold_legend(ax.get_legend())
ax.invert_xaxis()
annotate(ax, 600, 4000, axis="wn", ypos=0.88, fs=6.5)

# ── Panel 4: Attribution map ───────────────────────────────────────────────────
ax = ax_attr
abs_res = np.abs(residual_norm)
ax.fill_between(common_wn, 0, abs_res, color=RESID_COL, alpha=0.75,
                label="|Residual|  (H₂O + CH₃OH signature)")
ax.set_xlabel("Wavenumber (cm⁻¹)")
ax.set_ylabel("|Residual|")
ax.set_title("Attribution Map  —  Peaks = H₂O / CH₃OH ice features")
ax.legend(framealpha=0.85, facecolor=BG_PANEL, edgecolor=GRID_COL)
_bold_legend(ax.get_legend())
ax.invert_xaxis()
ylo7, yhi7 = ax.get_ylim()
for name,(lam,wn_c,_,col) in H2O_CH3OH.items():
    if 600 < wn_c < 4000:
        ax.axvline(wn_c, color=col, lw=0.9, alpha=0.70, ls=":")
        ax.text(wn_c, ylo7+0.88*(yhi7-ylo7), name, rotation=90,
                fontsize=6.5, color=col, va="top", ha="center", fontweight="bold")
ax.set_ylim(ylo7, yhi7)

save_fig(fig1, "H2O_CH3OH_v4w_Main")
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
# 14.  FIGURE 2 — ICE ZOOM
# ─────────────────────────────────────────────────────────────────────────────
fig2, axes2 = plt.subplots(1, 2, figsize=(16, 7))
fig2.patch.set_facecolor(BG_MAIN)
for ax in axes2: _style(ax)
fig2.suptitle(
    f"H₂O + CH₃OH Ice  —  Zoom  [{src_lbl}]  ·  Ice Absorption Detail",
    color=TEXT_COL, fontsize=13, fontweight="bold")

ax = axes2[0]
m = (wave_um >= 5.5) & (wave_um <= 13.0)
ax.plot(wave_um[m], flux_jy[m], color=SPEC_COL, lw=1.3)
ax.plot(wave_um[m], cont_jy[m], color=CONT_COL, lw=1.6, ls="--", alpha=0.85)
ax.fill_between(wave_um[m], flux_jy[m], cont_jy[m],
                where=(flux_jy[m] < cont_jy[m]), color=ABS_COL, alpha=0.28)
ax.fill_between(wave_um[m], flux_jy[m], cont_jy[m],
                where=(flux_jy[m] >= cont_jy[m]), color=EMI_COL, alpha=0.18)
ax.set_xlim(5.5, 13.0)
ax.set_xlabel("Wavelength (µm)")
ax.set_ylabel("Flux density (Jy)")
ax.set_title("FLUX  5.5–13 µm  —  H₂O bend 6.0  ·  CH₃OH 6.85  ·  Si+CH₃OH 9.7  (dips ↓)")
annotate(ax, 5.5, 13.0, axis="wl", ypos=0.90, fs=7.5)

ax = axes2[1]
m2 = (common_wn >= 800) & (common_wn <= 2000)
ax.plot(common_wn[m2], jwst_abs[m2], color=SPEC_COL, lw=1.3)
ax.fill_between(common_wn[m2], 0, jwst_abs[m2], color=ABS_COL, alpha=0.25)
ax.set_xlabel("Wavenumber (cm⁻¹)")
ax.set_ylabel("Normalised Absorbance [0–1]")
ax.set_title("ABSORBANCE  800–2000 cm⁻¹  —  same region  (peaks UP ↑ = ice absorption)")
ax.invert_xaxis()
annotate(ax, 800, 2000, axis="wn", ypos=0.90, fs=7.5)

plt.tight_layout()
save_fig(fig2, "H2O_CH3OH_v4w_Zoom")
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
# 15.  FIGURE 3 — CNN Training quality
# ─────────────────────────────────────────────────────────────────────────────
fig3, axes3 = plt.subplots(2, 3, figsize=(16, 9))
fig3.patch.set_facecolor(BG_MAIN)
fig3.suptitle("CNN Training Quality  —  H₂O + CH₃OH Lab Absorbance Spectra",
              color=TEXT_COL, fontsize=14, fontweight="bold")
for ax in axes3.flat: _style(ax)

axes3[0,0].plot(history.history["loss"],     color=SPEC_COL, lw=1.4, label="Train")
axes3[0,0].plot(history.history["val_loss"], color=CONT_COL, lw=1.4, ls="--", label="Val")
axes3[0,0].set_xlabel("Epoch"); axes3[0,0].set_ylabel("Weighted MSE Loss")
axes3[0,0].set_title("Training Loss"); axes3[0,0].legend(framealpha=0.85)
_bold_legend(axes3[0,0].get_legend())

axes3[0,1].plot(history.history["mae"],     color=EMI_COL,   lw=1.4, label="Train")
axes3[0,1].plot(history.history["val_mae"], color=RESID_COL, lw=1.4, ls="--", label="Val")
axes3[0,1].set_xlabel("Epoch"); axes3[0,1].set_ylabel("MAE")
axes3[0,1].set_title("Mean Absolute Error"); axes3[0,1].legend(framealpha=0.85)
_bold_legend(axes3[0,1].get_legend())

r2_each = [r2_score(X[i], X_recon[i]) for i in range(len(X))]
axes3[0,2].hist(r2_each, bins=max(5,len(r2_each)//3),
                color=TAU_COL, alpha=0.80, edgecolor="white")
axes3[0,2].axvline(np.mean(r2_each), color=RESID_COL, lw=2, ls="--",
                   label=f"Mean R²={np.mean(r2_each):.4f}")
axes3[0,2].set_xlabel("R²"); axes3[0,2].set_ylabel("Count")
axes3[0,2].set_title("Lab R² Distribution (original space)")
axes3[0,2].legend(framealpha=0.85)
_bold_legend(axes3[0,2].get_legend())

best_idx = np.argsort(r2_each)[-3:]
for k, idx in enumerate(best_idx):
    ax = axes3[1,k]; _style(ax)
    ax.plot(common_wn, X[idx],       color=TEXT_COL, lw=1.4, label="Original")
    ax.plot(common_wn, X_recon[idx], color=SPEC_COL, lw=1.4, ls="--",
            label=f"Recon R²={r2_each[idx]:.4f}")
    ax.set_xlabel("Wavenumber (cm⁻¹)"); ax.set_ylabel("Norm. Absorbance")
    ax.set_title(f"{names[idx][:28]}")
    ax.legend(framealpha=0.85, facecolor=BG_PANEL, edgecolor=GRID_COL)
    _bold_legend(ax.get_legend())
    ax.invert_xaxis()

plt.tight_layout()
save_fig(fig3, "H2O_CH3OH_v4w_Training")
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
# 16.  Summary
# ─────────────────────────────────────────────────────────────────────────────
print(f"\n{'═'*72}")
print(f"  ✅  COMPLETE  (v4-white)")
print(f"  Flux display  : {chosen_name}  [{src_lbl}]")
print(f"  CNN domain    : Synthetic H₂O+CH₃OH absorbance (lab mean + smooth boosts)")
print(f"  JWST R²       : {r2:.4f}  {'✓ TARGET MET (≥0.98)' if r2>=0.98 else '→ target 0.98'}")
print(f"  Pearson r     : {pearson_r:.4f}")
print(f"  Lab mean R²   : {r2_lab:.4f}")
print(f"  Figures       : {FIGURES_DIR}/")
print(f"{'═'*72}")
```
